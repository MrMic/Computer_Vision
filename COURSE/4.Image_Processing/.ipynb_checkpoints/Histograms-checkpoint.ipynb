```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

## PART 1

```python
dark_horse = cv2.imread('../../Computer-Vision-with-Python/DATA/horse.jpg') # ORIGINAL BGR OPENCV
show_horse = cv2.cvtColor(dark_horse,cv2.COLOR_BGR2RGB) # CONVERTED TO RGB FOR SHOW

rainbow = cv2.imread('../../Computer-Vision-with-Python/DATA/rainbow.jpg')
show_rainbow = cv2.cvtColor(rainbow,cv2.COLOR_BGR2RGB)

blue_bricks = cv2.imread('../../Computer-Vision-with-Python/DATA/bricks.jpg')
show_bricks = cv2.cvtColor(blue_bricks,cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(show_horse)
```

```python
# OPENCV BGR
hist_values = cv2.calcHist([blue_bricks],channels=[0],mask=None,histSize=[256],ranges=[0,256])
```

```python
hist_values.shape
```

```python
plt.plot(hist_values)
```

```python
# OPENCV BGR
hist_values = cv2.calcHist([dark_horse],channels=[0],mask=None,histSize=[256],ranges=[0,256])
plt.plot(hist_values)
```

```python
# OPENCV BGR
hist_values = cv2.calcHist([rainbow],channels=[0],mask=None,histSize=[256],ranges=[0,256])
plt.plot(hist_values)
```

```python
img = blue_bricks
```

```python
color = ('b','g','r')

for i, col in enumerate(color):
    histr = cv2.calcHist([img],[i],mask=None,histSize=[256],ranges=[0,256])
    plt.plot(histr,color=col)
    plt.xlim([0,256])

plt.title('Histogramme Blue Bricks')
```

```python
img = dark_horse
color = ('b','g','r')

for i, col in enumerate(color):
    histr = cv2.calcHist([img],[i],mask=None,histSize=[256],ranges=[0,256])
    plt.plot(histr,color=col)
    plt.xlim([0,256])
    plt.ylim([0,100000])

plt.title('Histogramme Dark Horse')
```

## PART 2

```python
rainbow = cv2.imread('../../Computer-Vision-with-Python/DATA/rainbow.jpg')
show_rainbow = cv2.cvtColor(rainbow,cv2.COLOR_BGR2RGB)
```

```python
img = rainbow
```

```python
img.shape
```

```python
mask = np.zeros(img.shape[:2],np.uint8)
```

```python
mask[300:400,100:400] = 255
```

```python
plt.imshow(mask,cmap='gray')
```

```python
masked_img = cv2.bitwise_and(img,img,mask=mask)
```

```python
# plt.imshow(masked_img)
show_masked_img = cv2.bitwise_and(show_rainbow,show_rainbow,mask=mask)
```

```python
plt.imshow(show_masked_img)
```

```python
# OPENCV => B G R
hist_mask_values_red = cv2.calcHist([rainbow],channels=[2],mask=mask,histSize=[256],ranges=[0,256])
```

```python
hist_values_red = cv2.calcHist([rainbow],channels=[2],mask=None,histSize=[256],ranges=[0,256])
```

```python
plt.plot(hist_mask_values_red)
plt.title('RED HISTGRAM FOR MASKED RAINBOW')
```

```python
plt.plot(hist_values_red)
plt.title('RED HISTGRAM FOR NORMAL RAINBOW')
```
